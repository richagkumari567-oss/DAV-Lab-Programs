```python
import pandas as pd
import matplotlib.pyplot as plt#to plot graphs
import seaborn as sns
print("LIBRARIES IMPORTED SUCCESSFULLY!")
```

seaborn and matplotlib are python libraries used for visualiztion.both are used to plot a graph but we use seaborn for creative graph(advanced)

```python
data={"Student":["Asha","Bharat","Chitra","Deepak","Esha","Farhan",
                 "Gowri","Harish","Isha","Jatin","Kavya","Lokesh",
                 "Meera","Nikhil","Pooja","Rahul","Sneha","Tarun","Uma","Varun"],
      "Department":["CSE","AIML","CSE","ISE","AIML","CSE","ISE"
      ,"CSE","AIML","ISE","CSE","AIML",
                    "ISE","CSE","AIML","ISE","CSE","AIML","ISE","CSE"],
      "Attendance":[92,78,85,96,88,72,91,65,87,82,94,76,89,97,84,69,93,80,86,90],
      "Study_Hours":[3.5,2,3,5,4,1.5,4.5,1,3.5,2.5,5.5,2,4,6,3,1.5,5,2.5,3.5,4.5],
      "Assignment_Score":[82,68,78,94,86,62,90,55,84,73,96,70,88,98,79,60,93,74,85,91],
      "Internal_Marks":[84,70,80,95,87,65,91,58,85,75,97,72,89,99,81,63,94,76,86,92],
      "Final_Score":[86,72,82,96,88,66,93,60,86,77,98,73,90,99,83,64,95,78,87,94]}
df=pd.DataFrame(data)
display(df.head())
print("Shape:",df.shape)
print(df.dtypes)
```

figure size is heightand width.bins means columns,edge color is beans color(border color),title is title of graph,x label is x axix name,seaborn makes statistical analysis easier.

```python
plt.figure(figsize=(3,3))
plt.hist(df["Final_Score"],bins=6,edgecolor="black")
plt.title("Distribution of final scores")
plt.xlabel("Final score");plt.ylabel("Number of students")
plt.show()
```

```python
plt.figure(figsize=(3,3))
plt.hist(df["Final_Score"],bins=10,edgecolor="black")
plt.title("Distribution of final scores")
plt.xlabel("Final score");plt.ylabel("Number of students")
plt.show()
```

```python
plt.figure(figsize=(3,3))
sns.boxplot(y=df["Final_Score"])
plt.title("Box plot of final score")
plt.show()
```

```python
plt.figure(figsize=(3,3))
sns.boxplot(data=df,x="Department",y="Final_Score")
plt.title("Final Score distribution by department")
plt.show()
```

```python
#scatter plot
plt.figure(figsize=(3,3))
sns.scatterplot(data=df,x="Attendance",y="Final_Score")
plt.title("Scatter plot of attendance vs final score")
plt.show()
```

```python
plt.figure(figsize=(3,3))
sns.scatterplot(data=df,x="Study_Hours",y="Final_Score",hue="Department",s=100)
plt.title("Scatter plot of study hours vs final score")
plt.show()
```

```python
num_cols=["Attendance","Study_Hours","Assignment_Score","Internal_Marks","Final_Score"]
plt.figure(figsize=(3,3))
sns.pairplot(df[num_cols])
plt.show()
#numerical columns which has only numbers, it will plot separete graph for each numerical variables
```

```python
corr=df[num_cols].corr()
plt.figure(figsize=(3,3))
#sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")
#sns.heatmap(corr,annot=True,cmap="bwr",fmt=".2f")
sns.heatmap(corr,annot=True,cmap="viridis",fmt=".2f")
plt.title("Correlation matrix")
plt.show()
#display(corr)
```
