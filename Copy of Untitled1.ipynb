```python
import numpy as np
import pandas as pd
print("Numpy version:",np.__version__)
print("Pandas version:",pd.__version__)
```

```python
marks=np.array([78,85,92,67,88])
print("Marks:",marks)
print("Type:",type(marks))
```

```python
print("Number of dimensions:",marks.ndim)
print("shape:",marks.shape)
print("Number of elements:",marks.size)
```

```python
print("Original:",marks)
print("Add 5:",marks+5)
print("Multiply by 2:",marks*2)
print("Divide by 2:",marks/2)
print("Square:",marks**2)
```

```python
print("SUM",marks.sum())
print("MEAN",marks.mean())
print("MAXIMUM",marks.max())
print("MINIMUM",marks.min())
```

```python
print("First mark:",marks[0])
print("second mark:",marks[1])
print("last mark:",marks[-1])
```

```python
print("Index 1 to 3:",marks[1:4])
print("from start to 4:",marks[:4])
print("from 2 to end:",marks[2:])
print("every second element:",marks[::2])
```

```python
student_mark=np.array([
    [78,85,92],
    [67,88,92],
    [90,95,89]
])
print(student_mark)
print("Shape:",student_mark.shape)
```

```python
print("Mean of all values:",student_mark.mean())
print("Mean of each column:",student_mark.mean(axis=0))
print("Mean of each row:",student_mark.mean(axis=1))
```
